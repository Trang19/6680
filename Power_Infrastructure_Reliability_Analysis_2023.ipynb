```python
import pandas as pd

# File paths
eagle_file = "/content/eaglei_outages_with_events_2023_24_hours_lag.csv"
noaa_file = "/content/StormEvents_details-ftp_v1.0_d2023_c20260323.csv"

# Load EAGLE-I outage data
eagle = pd.read_csv(eagle_file)

# Load NOAA Storm Events data
noaa = pd.read_csv(noaa_file, low_memory=False)

print("EAGLE-I shape:", eagle.shape)
print("NOAA shape:", noaa.shape)
```

```python
print("EAGLE-I columns:")
print(eagle.columns.tolist())

print("\nNOAA columns:")
print(noaa.columns.tolist())
```

```python
print("\nFirst 5 EAGLE-I rows:")
display(eagle.head())

print("\nFirst 5 NOAA rows:")
display(noaa.head())
```

```python
# Check the exact data types
print("EAGLE-I data types:")
print(eagle[['event_id', 'Event Type', 'fips', 'start_time',
             'duration', 'min_customers', 'max_customers',
             'mean_customers']].dtypes)

print("\nNOAA important columns:")
print(noaa[['EVENT_ID', 'STATE', 'STATE_FIPS', 'CZ_FIPS',
            'CZ_TYPE', 'EVENT_TYPE', 'BEGIN_YEARMONTH',
            'BEGIN_DAY', 'BEGIN_TIME',
            'END_YEARMONTH', 'END_DAY', 'END_TIME',
            'MAGNITUDE']].dtypes)
```

```python
print("NOAA CZ_TYPE values:")
print(noaa['CZ_TYPE'].value_counts(dropna=False))

print("\nNOAA EVENT_TYPE values:")
print(noaa['EVENT_TYPE'].value_counts().head(30))
```

```python
print("EAGLE-I Event Types:")
print(eagle['Event Type'].value_counts())
```
